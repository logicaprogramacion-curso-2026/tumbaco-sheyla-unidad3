# Tema: NumPy



**¿Qué es NumPy?**

Numerical Python es una biblioteca de Python especializada en el cálculo y el manejo de arreglos multidimensionales. Fue diseñada para realizar operaciones matemáticas de manera rápida y eficiente, siendo una de las herramientas más utilizadas en ciencia de datos, inteligencia artificial, ingeniería y análisis de información. Su principal característica es el uso de los arreglos, los cuales permiten almacenar cantidades de datos y realizar operaciones sobre todos sus elementos al mismo tiempo, sin necesidad de utilizar múltiples ciclos.

**¿Para qué sirve?**

Se utiliza para facilitar el procesamiento de datos numéricos. Entre sus principales aplicaciones están el crear vectores y matrices, realizar operaciones matemáticas entre arreglos, resolver cálculos estadísticos, manipular datos multidimensionales, trabajar con álgebra linea, generar números aleatorios y servir como base para otras bibliotecas como Pandas, SciPy, TensorFlow y Scikit-Learn.

Gracias a su optimización, NumPy ejecuta operaciones mucho más rápido que las lsitas tradicionales de Pyhton.

```python
# Ejemplo 1
import numpy as np

notas = np.array([9, 8, 10, 7, 9])

promedio = np.mean(notas)

print("Notas:", notas)
print("Promedio:", promedio)


# Explicación: se crear un arreglo con cinco calificaciones y posteriormente se utiliza la función mean () para calcular automáticamente el promedio de todos los valores.
```

```python
# Ejemplo 2
import numpy as np

A = np.array([[2, 4],
              [6, 8]])

B = np.array([[1, 3],
              [5, 7]])

suma = A + B
multiplicacion = A * B
print("Matriz A")
print(A)

print("Matriz B")
print(B)

print("Suma")
print(suma)

print("Multiplicación elemento por elemento")
print(multiplicacion)
```
